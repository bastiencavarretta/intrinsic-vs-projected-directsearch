# Data Profiles 

```python
expnumber = 1
maniftypeobj = [(1, 2), (1, 1), (1, 3), (2, 1)]
saving = True

for rotation in [0, 1]:
    for lbdfoeuclideansimplex in [0, 1]:
        if lbdfoeuclideansimplex == 1:
            euclideansimplex = 1
            plotting_dp(
                expnumber,
                maniftypeobj=maniftypeobj,
                rotations=[rotation],
                psstypes=[1, 2, 3],
                projections=[0, 1],
                euclideansimplex=euclideansimplex,
                lbdfoeuclideansimplex=lbdfoeuclideansimplex,
                tau=1e-2,
                N=100,
                plotcodimsev=True,
                plotmdimsev=True,
                saving=saving,
            )
        euclideansimplex = 0
        plotting_dp(
            expnumber,
            maniftypeobj=maniftypeobj,
            rotations=[rotation],
            psstypes=[1, 2, 3],
            projections=[0, 1],
            euclideansimplex=euclideansimplex,
            lbdfoeuclideansimplex=lbdfoeuclideansimplex,
            tau=1e-2,
            N=100,
            plotcodimsev=True,
            plotmdimsev=True,
            saving=saving,
        )
```
